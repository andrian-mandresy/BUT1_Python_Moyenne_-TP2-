# Calcul des moyennes des UE (Version 1)

# Algo pour le calcul des moyennes des UE :
1°: Demander à l'utilisateur ses notes pour chaque modules des trois UE

2°: Demander à l'utilisateur ensuite les coefficients de chaque modules

3°: Mettre en place un calcul de ces notes multipliés par chaque coéfficients puis diviser par la somme des coéfficients

4°: Si la moyenne sur l'année de chacune des trois UE est ≥10, afficher : "l'année est validée et le passage en BUT 2 est autorisé"

5°: Si la moyenne sur l'année de 2 UE est ≥10, alors que la moyenne de la troisième UE est comprise entre 8 (inclus) et 10 (exclu), afficher : "l'année de BUT 1 n'est pas validée mais le passage en BUT 2 est autorisé. Il faudra alors impérativement valider cette dernière UE en BUT 2 (au niveau 2, ce qui validera automatiquement le niveau 1 )"

6°: Si la moyenne sur l'année de 2 UE est ≥10, alors que la moyenne de la troisième UE est inférieure à 8 : "l'année n'est pas validée et le passage en BUT 2 n'est pas autorisé. Un redoublement peut-être proposé à l'appréciation du conseil des enseignants."

7°: Si la moyenne sur l'année de 2 ou 3 UE est inférieure à 10 : "l'année n'est pas validée et le passage en BUT2 est exclu."

```python
def calcul_moy():
    sommeR1=0 # on initialise la somme des notes pour les modules R à 0
    somme_coeffR1=0 # on initialise la somme des coeffs pour les modules R à 0
    sommeSAE1=0 # on initialise la somme des notes pour les modules SAE à 0
    somme_coeffSAE1=0 # on initialise la somme des coeffs pour les modules SAE à 0
    for i in range (101,104): # on crée une boucle pour les modules R
        noteR1=float(input("Donnez votre note pour le module R{} pour l'UE1 :".format(i))) # on demande à l'utilisateur ses notes pour chaque module de R
        coeffR1=int(input("Donnez le coéfficient pour le module R{} pour l'UE1 :".format(i))) # on demande à l'utilisateur ses coeffs pour chaque module de R
        sommeR1=sommeR1+noteR1*coeffR1 # ici on multiplie chaque note par son coeff saisie par l'utilisateur
        somme_coeffR1=somme_coeffR1+coeffR1 # ici on additionne chaque coeff saisie par l'utilisateur
    for s in range (11,14): # on crée une boucle pour les modules R
        noteSAE1=float(input("Donnez votre note pour le module SAE{} pour l'UE1 :".format(s))) # on demande à l'utilisateur ses notes pour chaque module de SAE
        coeffSAE1=int(input("Donnez le coéfficient pour le module SAE{} pour l'UE1 :".format(s))) # on demande à l'utilisateur ses coeffs pour chaque module de SAE
        sommeSAE1=sommeSAE1+noteSAE1*coeffSAE1 # ici on multiplie chaque note par son coeff saisie par l'utilisateur
        somme_coeffSAE1=somme_coeffSAE1+coeffSAE1 # ici on additionne chaque coeff saisie par l'utilisateur
    moy_1=(sommeR1+sommeSAE1)/(somme_coeffR1+somme_coeffSAE1) # enfin on divise la somme des notes en R et SAE sur la somme des coeffs en R et SAE
    print(moy_1)
    
    sommeR2=0
    somme_coeffR2=0
    sommeSAE2=0
    somme_coeffSAE2=0
    for i in range (101,104):
        noteR2=float(input("Donnez votre note pour le module R{} pour l'UE2 :".format(i)))
        coeffR2=int(input("Donnez le coéfficient pour le module R{} pour l'UE2 :".format(i)))
        sommeR2=sommeR2+noteR2*coeffR2
        somme_coeffR2=somme_coeffR2+coeffR2
    for s in range (11,14):
        noteSAE2=float(input("Donnez votre note pour le module SAE{} pour l'UE2 :".format(s)))
        coeffSAE2=int(input("Donnez le coéfficient pour le module SAE{} pour l'UE2 :".format(s)))
        sommeSAE2=sommeSAE2+noteSAE2*coeffSAE2
        somme_coeffSAE2=somme_coeffSAE2+coeffSAE2
    moy_2=(sommeR2+sommeSAE2)/(somme_coeffR2+somme_coeffSAE2)
    print(moy_2)

    sommeR3=0
    somme_coeffR3=0
    sommeSAE3=0
    somme_coeffSAE3=0
    for i in range (101,104):
        noteR3=float(input("Donnez votre note pour le module R{} pour l'UE3 :".format(i)))
        coeffR3=int(input("Donnez le coéfficient pour le module R{} pour l'UE3 :".format(i)))
        sommeR3=sommeR3+noteR3*coeffR3
        somme_coeffR3=somme_coeffR3+coeffR3
    for s in range (11,14):
        noteSAE3=float(input("Donnez votre note pour le module SAE{} pour l'UE3 :".format(s)))
        coeffSAE3=int(input("Donnez le coéfficient pour le module SAE{} pour l'UE3 :".format(s)))
        sommeSAE3=sommeSAE3+noteSAE3*coeffSAE3
        somme_coeffSAE3=somme_coeffSAE3+coeffSAE3
    moy_3=(sommeR3+sommeSAE3)/(somme_coeffR3+somme_coeffSAE3)
    print(moy_3)


calcul_moy()

```

# Calcul des moyennes des UE avec listes (Version 2)

```python
import numpy as np

# création de la liste de toutes les matières (ressources et SAE)
matière=[f"R{i}" for i in range (101,116)]+[f"SAE{i}" for i in range(11,17)]


"""
# récupération des notes de toutes les matières
note=[float(input(f"Quelle est la note pour {elt} ?")) for elt in matière]
"""

# c'est pour générer des notes allant de 0 à 20 : 
note=[i for i in range(21)]

# création des listes de coéfficients
coeff1=[10,10,7,7,0,5,0,6,0,5,4,2,5,5,0,20,20,0,0,0,7]
coeff2=[4,0,2,8,6,0,0,0,0,5,5,2,9,9,3,0,0,29,0,0,7]
coeff3=[4,0,2,0,0,5,15,6,4,5,5,2,0,0,3,0,0,0,20,20,7]

# round c'est pour les nombres après la virgule, averrage c'est pour la moyenne, weights c'est pour les poids
def moy(coeff):
    moy=round(np.average(note,weights=coeff),3)
    return moy
print (f"Les moyennes des UE sont respectivement UE1 {moy(coeff1)}, UE2 {moy(coeff2)} et UE3 {moy(coeff3)}")
```
